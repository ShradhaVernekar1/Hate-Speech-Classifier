```python
#creating a new label to classify the tweets
dataset["labels"]=dataset["class"].map({0:"Hate",1:"Offensive",2:"no hate or offensive"})
dataset
```

```python
data=dataset[["tweet","labels"]]
data
```

```python
#for pre-processing
import re #to get rid of special symbols from tweets
import nltk #removal of stop words and stemming words
import string
```

```python
nltk.download('stopwords')
```

```python
#importing stop words
from nltk.corpus import stopwords
stopwords = set(stopwords.words("english"))
```

```python
#importing Stemmer
stemmer=nltk.SnowballStemmer("english")
```

```python
#Data Cleaning
def clean_data(text):
  text=str(text).lower()
  text=re.sub('http?://\S+|www\.S+','',text)
  text=re.sub('\[.*?\]','',text)
  text=re.sub('<.*?>+','',text)
  text=re.sub('[%s]'%re.escape(string.punctuation),'',text)
  text=re.sub('\n','',text)
  text=re.sub('\w*\d\w*','',text)
  #stopword removal
  text=[word for word in text.split(' ') if word not in stopwords]
  text=" ".join(text)
  #stemming the text
  text=[stemmer.stem(word) for word in text.split(' ')]
  text=" ".join(text)
  return text
```

```python
data['tweet']= data['tweet'].apply(clean_data)
```

```python
data
```

```python
X=np.array(data["tweet"])
Y=np.array(data["labels"])
```

```python
from sklearn.feature_extraction.text import CountVectorizer #to convert array into vectors
from sklearn.model_selection import train_test_split # to train n test the dataset
```

```python
cv=CountVectorizer() #creating a vector object
X=cv.fit_transform(X)
X
```

```python
#traing  n testing
X_train, X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.33,random_state=42)
```

```python
#building the model
from sklearn.tree import DecisionTreeClassifier
```

```python
dt=DecisionTreeClassifier()
dt.fit(X_train,Y_train)
```

```python
Y_pred=dt.predict(X_test)
```

```python
#confusion matrix n accuracy
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(Y_test,Y_pred)
cm
```

```python
import seaborn as sns
import matplotlib.pyplot as ply
%matplotlib inline
```

```python
sns.heatmap(cm, annot=True, fmt="d", cmap="YlGnBu")
ply.xlabel("Predicted")
ply.ylabel("Actual")
ply.title("Confusion Matrix")
ply.show()
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(Y_test,Y_pred)
```
